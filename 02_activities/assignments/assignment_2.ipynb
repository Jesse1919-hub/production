### Preproc 1

Create preproc1 below.

+ Numeric: scaled variables, no other transforms.
+ Categorical: one-hot encoding.

```python
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, RobustScaler, OneHotEncoder, PowerTransformer

# Data features
numerical_features = ['coord_x', 'coord_y', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain']   
categorical_features = ['month', 'day'] 

# Preprocessing for Prteproc 1
preproc1 = ColumnTransformer(
    transformers=[(
        'num', StandardScaler(), numerical_features
    ), (
        'cat', OneHotEncoder(handle_unknown='ignore'), categorical_features
    )])
```

### Preproc 2

Create preproc1 below.

+ Numeric: scaled variables, non-linear transformation to one or more variables.
+ Categorical: one-hot encoding.

```python
preproc2 = ColumnTransformer(
       transformers=[(
        'num_non_linear', PowerTransformer(method='yeo-johnson'), numerical_features
    ), (
        'cat', OneHotEncoder(handle_unknown='ignore'), categorical_features
    )])
```

# Tune Hyperparams

+ Perform GridSearch on each of the four pipelines. 
+ Tune at least one hyperparameter per pipeline.
+ Experiment with at least four value combinations per pipeline.

```python
from sklearn.model_selection import GridSearchCV

# simple model hyperparameter grid
param_grid_simple = {
    'regressor__n_neighbors': [3, 5, 7, 9],
    'regressor__weights': ['uniform', 'distance'],
}

# advanced model hyperparameter grid
param_grid_advanced = {
    'regressor__n_estimators': [100, 200],
    'regressor__max_depth': [None, 10, 20],
    'regressor__min_samples_split': [2, 5],
}
```

```python
pipelines = {
    'pipeline_A': (pipeline_A, param_grid_simple),  
    'pipeline_B': (pipeline_B, param_grid_simple),
    'pipeline_C': (pipeline_C, param_grid_advanced),
    'pipeline_D': (pipeline_D, param_grid_advanced),
}
```

```python
best_models = {}
results = {}
for name, (pipeline, param_grid) in pipelines.items():
    grid_search = GridSearchCV(estimator=pipeline, param_grid=param_grid, cv=5, n_jobs=-1, scoring='neg_mean_squared_error')
    grid_search.fit(x, Y)

    best_models[name] = grid_search.best_estimator_
    results[name] = grid_search.cv_results_
    print(f"Best parameters for {name}: {grid_search.best_params_}, negative MSE: {grid_search.best_score_}")
```
